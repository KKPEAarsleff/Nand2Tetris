```python
input_file = 'Prog.asm'

output_file = input_file.split('.')[0] + '.hack'

assembly_lines = []

parsed_lines = []
```

```python

def assemble(input_file):
        
        with open(input_file, 'r') as infile:
            for line in infile:
                stripped_line = line.strip().split("//")[0].strip() # removing comments and whitespaces
                if stripped_line:
                     assembly_lines.append(stripped_line) # append to list if not blank
                

        with open (output_file, 'w') as outfile:
             for line in assembly_lines:
                  outfile.write(line + "\n") # write to hack program file
```

```python
comp = {'0': '101010',
         '1': '111111',
         '-1': '111010',
         'D': '001100',
         'A': '110000',
         '!D': '001101',
         '!A': '110001',
         '-D': '001111',
         '-A': '110011',
         'D+1': '011111',
         'A+1': '110111',
         'D-1': '001110',
         'A-1': '110010',
         'D+A': '000010',
         'D-A': '010011',
         'A-D': '000111',
         'D&A': '000000',
         'D|A': '010101'}
```

```python
dest = {'null': '000',
        'M': '001',
        'D': '010',
        'DM': '011',
        'A': '100',
        'AM': '101',
        'AD': '110',
        'ADM': '111'}
```

```python
jump = {'null': '000',
        'JGT': '001',
        'JEQ': '010',
        'JGE': '011',
        'JLT': '100',
        'JNE': '101',
        'JLE': '110',
        'JMP': '111'}
```

```python
def parse(assembly_lines):
    for line in assembly_lines:
        if line[0] == '@':
            value = bin(int(line[1:]))[2:].zfill(16)
            parsed_lines.append(value)
        elif '=' in line:
            dest_part, comp_part = line.split('=')
            value = '1111' + comp[comp_part] + dest[dest_part] + '000'
            parsed_lines.append(value) 
        elif ';' in line:
            comp_part, jump_part = line.split(';')
            value = '1111' + comp[comp_part] + '000' + jump[jump_part]
            parsed_lines.append(value)
        
```

```python
assemble(input_file)
```

```python
def write_to_executeable(parsed_lines, output_file):
    with open(output_file, 'w') as outfile:
        for line in parsed_lines:
            outfile.write(line + "\n")
```
